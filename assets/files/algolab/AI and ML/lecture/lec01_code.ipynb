## Slide 19 — The negative gradient points downhill

Try one-dimensional gradient steps first.

```python
### notes
# the gradient tells us which way is up
# parameter ~ weighs how important each feature is, it is learnable (change it based on the loss)
# hyperparameter ~ constant - we just choose a value and never change it during the training.

# theta - parameter
# eta - step size (hyperparameter)


# v1 @ v1 === norm(v1) ^ 2
# v1 = [a, b], norm(v1) = sqrt(a**2 + b**2)
# v1 @ v1 = a**2 + b**2
# rT r = r @ r = square(norm(r))

# mean square error (MSE)
# the average error = the average residual square
# 1/n ||r || ^ 2


# v1 = [[a], [b]] 2 x 1
# v2 = [[c], [d]] (2 x 1)
# v1 @ v2 = ac + bd
# v1.T = [[a, b]] (1 x 2)
# v1.T  v2  (1x1) = [ac + bd]
# v1 @ v2 = v1.T v2
```

```python
# Prediction: For L(w)=w² at w=4, what is the loss?
# My prediction:
#

w_value = 4
w_value ** 2
```

```python
# Prediction: What is the gradient at w=4?
# My prediction:
#

gradient_value = 2 * w_value
gradient_value
```

```python
# Prediction: With step size 0.1, what is the new w?
# My prediction:
#

new_w = w_value - 0.1 * gradient_value
new_w
```

```python
# Prediction: What is the new loss?
# My prediction:
#

new_w ** 2
```

```python
# Prediction: What happens with a very large step size 2?
# My prediction:
#

large_step_w = w_value - 2 * gradient_value
large_step_w ** 2
```

```python
# Prediction: Where does the small update move on the loss curve?
# My prediction:
#

w_grid = np.linspace(-5, 5, 200)
plt.plot(w_grid, w_grid ** 2, label='L(w) = w²')
plt.scatter([w_value, new_w], [w_value**2, new_w**2], color=['orange', 'purple'])
plt.annotate('', xy=(new_w, new_w**2), xytext=(w_value, w_value**2),
             arrowprops={'arrowstyle': '->', 'color': 'purple'})
plt.xlabel('w'); plt.ylabel('loss'); plt.legend(); plt.show()
```

<details>
<summary><strong>click to see the solutions</strong></summary>

The small step moves `w` from `4` to `3.2`, lowering loss from `16` to `10.24`. The plot locates that update on the loss curve. The very large step increases loss to `144`; direction and step size both matter.

</details>

## Slide 29 — Complexity creates underfitting and overfitting

Use the same points with a straight line and a flexible curve. `np.polyfit` asks Python to fit a polynomial; you do not need to memorize it.

```python
# Prediction: How many training points are there?
# My prediction:
#

x_curve = np.linspace(-1, 1, 8)
y_curve = np.sin(np.pi * x_curve)
len(x_curve)
```

```python
# Prediction: What is the straight-line model's training error?
# My prediction:
#

line = np.polyfit(x_curve, y_curve, 1)
line_predictions = np.polyval(line, x_curve)
np.mean((line_predictions - y_curve) ** 2)
```

```python
# Prediction: What is the degree-7 model's training error?
# My prediction:
#

flexible = np.polyfit(x_curve, y_curve, 7)
flexible_predictions = np.polyval(flexible, x_curve)
np.mean((flexible_predictions - y_curve) ** 2)
```

<details>
<summary><strong>click to see the solutions</strong></summary>

There are 8 points. The straight line underfits. The degree-7 polynomial can fit these noiseless points almost exactly; with noisy data it can also memorize noise.

</details>

## Slide 30 — Validation error reveals overfitting

Now add noise and compare models on new points.

```python
# Prediction: What are the noisy training targets?
# My prediction:
#

noise = np.array([0.25, -0.30, 0.20, -0.15, 0.10, -0.20, 0.30, -0.25])
y_noisy = y_curve + noise
y_noisy
```

```python
# Prediction: Fit degree 3. What is its validation error?
# My prediction:
#

useful = np.polyfit(x_curve, y_noisy, 3)
x_new = np.linspace(-0.9, 0.9, 20)
y_new = np.sin(np.pi * x_new)
np.mean((np.polyval(useful, x_new) - y_new) ** 2)
```

```python
# Prediction: Fit degree 7. What is its validation error?
# My prediction:
#

flexible = np.polyfit(x_curve, y_noisy, 7)
np.mean((np.polyval(flexible, x_new) - y_new) ** 2)
```

<details>
<summary><strong>click to see the solutions</strong></summary>

Degree 3 has the lower validation error here. Degree 7 fits the training points exactly but follows their noise between observations.

</details>

## Slide 31 — Optimization is not generalization

Low training error and low validation error answer different questions.

```python
# Prediction: Which model has lower training error?
# My prediction:
#

useful_train_error = np.mean((np.polyval(useful, x_curve) - y_noisy) ** 2)
flex_train_error = np.mean((np.polyval(flexible, x_curve) - y_noisy) ** 2)
useful_train_error, flex_train_error
```

```python
# Prediction: Which model has lower validation error?
# My prediction:
#

useful_validation_error = np.mean((np.polyval(useful, x_new) - y_new) ** 2)
flex_validation_error = np.mean((np.polyval(flexible, x_new) - y_new) ** 2)
useful_validation_error, flex_validation_error
```

<details>
<summary><strong>click to see the solutions</strong></summary>

Degree 7 wins on the fitted points. Degree 3 wins on validation data. Training success does not guarantee generalization.

</details>
