## 2. A model is a function with parameters

A one-feature linear model is

$$\widehat y=wx+b.$$

- The **input** $x$ describes one example.
- The **prediction** $\widehat y$ is the model's output.
- The **weight** $w$ controls how the prediction changes with the input.
- The **bias** $b$ shifts every prediction.
- The weight and bias are the adjustable **parameters**.

NumPy lets one equation predict an entire batch.

## 3. Residuals become a loss

For each example:

$$r_i=\widehat y_i-y_i.$$

- Positive residual: prediction is too high.
- Negative residual: prediction is too low.
- Residual size: amount of the miss.

Mean squared error summarizes all mistakes:

$$\operatorname{MSE}=\operatorname{mean}(r^2).$$

NumPy pattern: **subtract → square → mean**.

## 4. A gradient gives parameter directions

For a one-feature model:

$$g_w=\operatorname{mean}(2r_ix_i),\qquad
g_b=\operatorname{mean}(2r_i).$$

Read the weight formula as NumPy steps:

1. multiply each residual by its input;
2. multiply by 2;
3. average the contributions.

A positive gradient means increasing the parameter increases loss nearby.
A negative gradient means increasing it decreases loss nearby.

## 5. Gradient descent repeats the update

The update rule is

$$\text{new parameter}
=\text{old parameter}-\eta\cdot\text{gradient}.$$

The learning rate $\eta$ controls step size:

- too small: safe but slow;
- useful: steady progress;
- too large: overshooting or divergence.

The gradient must be recalculated after every update.

### Worked example 5A — trace three training steps

Run three complete prediction-loss-gradient-update cycles.

Try it in the scratch cell before opening the answer.

```python
example_5a_x = np.array([1.0, 2.0, 3.0])
example_5a_y = np.array([2.0, 3.0, 5.0])

# Write your own three-step loop before opening the answer.
```

<details>
        <summary><strong>Click for solution and explanation</strong></summary>

        <pre><code>weight = 1.0
bias = 0.0
learning_rate = 0.1

for step in range(3):
    predictions = weight * example_5a_x + bias
    residuals = predictions - example_5a_y
    loss = np.mean(residuals ** 2)
    gradient_weight = np.mean(2 * residuals * example_5a_x)
    gradient_bias = np.mean(2 * residuals)

    print(step, &quot;loss =&quot;, loss, &quot;w =&quot;, weight, &quot;b =&quot;, bias)

    weight -= learning_rate * gradient_weight
    bias -= learning_rate * gradient_bias</code></pre>

        <p>Each iteration uses the current parameters, recalculates the residuals and gradients, and only then updates. Reusing the first gradient forever would not be gradient descent.</p>
        </details>

## 6. Two features use matrix multiplication

One house with two features is a vector:

$$\widehat y=x^\top w+b.$$

A batch of houses is a matrix:

$$\widehat y=Xw+b.$$

Shapes:

- $X$: `(number of houses, number of features)`;
- $w$: `(number of features,)`;
- predictions: `(number of houses,)`.

**Fit** changes the parameters. **Predict** uses fixed parameters.

## Final learning trace

            Before finishing, point to the exact NumPy expression for each step:

            1. **Input:** the array or matrix holding the examples.
            2. **Prediction:** multiplication plus bias.
            3. **Residual:** prediction minus target.
            4. **Loss:** mean of squared residuals.
            5. **Gradient:** residuals combined with inputs.
            6. **Update:** old parameter minus learning rate times gradient.

            <details>
            <summary><strong>Click for the complete two-feature pattern</strong></summary>

            <pre><code>predictions = X @ weights + bias
residuals = predictions - y
loss = np.mean(residuals ** 2)

gradient_weights = (2 / len(X)) * X.T @ residuals
gradient_bias = 2 * np.mean(residuals)

weights = weights - learning_rate * gradient_weights
bias = bias - learning_rate * gradient_bias</code></pre>

            </details>

            understand the core of gradient descent.
